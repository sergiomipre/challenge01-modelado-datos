```python
import pandas as pd
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(10,5))

ax=df1.groupby("Tienda")['Precio'].sum().plot(kind='bar',title='Ingreso por Tienda')
for container in ax.containers:
  ax.bar_label(container,
               #fmt='$%.0f',
               fmt='${:,.0f}',
               label_type='edge')

#ax=plt.ticklabel_format(style='plain', axis='y') # Desactiva la notación científica

# Crear etiquetas personalizadas
#'{:,}'.format(int(x))
nuevas_etiquetas = [f'{i:,.0f}' for i in ax.get_yticks()]
ax.set_yticklabels(nuevas_etiquetas)

plt.xlabel('Tienda')
plt.ylabel('Pesos')
plt.show()
```

```python
ax= df1.groupby("Tienda")['Precio'].sum().sort_values(ascending=True).plot(kind='barh',title='Ingreso por Tienda',color=['skyblue','lightcoral','lightgreen','purple'])
for container in ax.containers:
  ax.bar_label(container, fmt='${:,.0f}', label_type='edge')

plt.ylabel('Tienda')
plt.xlabel('Pesos')
plt.show()
```

```python
ax= df1.groupby("Tienda")['Precio'].sum().sort_index(ascending=False).plot(kind='barh',title='Ingreso por Tienda',color=['skyblue','lightcoral','lightgreen','purple'])
for container in ax.containers:
  ax.bar_label(container, fmt='${:,.0f}', label_type='edge')

plt.ylabel('Tienda')
plt.xlabel('Pesos')
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(8, 6))
sns.barplot(data=df2, x=df2["Tienda"], y=df2['Ingreso por Ventas'],capsize=0.2, ax=ax, errorbar=None)

# show valor
for p in ax.patches:
    h, w, x = p.get_height(), p.get_width(), p.get_x()
    #xy = (x + w / 2., h / 2)
    xy = (x+w/2, h)
    text = f'${h:,.0f}'
    ax.annotate(text=text, xy=xy, ha='center', va='top')

plt.show()
```
